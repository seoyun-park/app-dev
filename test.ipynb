```python
call_url = "https://api.open-meteo.com/v1/forecast?latitude=36.8&longitude=127.1&current=temperature_2m,relative_humidity_2m&timezone=Asia/Seoul"
call_url
```

```python
# requests (1번 방식)
import requests
response = requests.get(call_url) #sync

response.json()
```

```python
# httpx (2번 방식)

only_url = "https://api.open-meteo.com/v1/forecast"
params_value = {
        "latitude": 36.8,
        "longitude": 127.1,
        "current": "temperature_2m,relative_humidity_2m",
        "timezone": "Asia/Seoul",
    }


params_value


import requests
response = requests.get(only_url,
                        params=params_value,
                        timeout=5) #sync


response.json()
```

```python
# httpx

import httpx

async def get_weather():
    async with httpx.AsyncClient(timeout=5) as client:
        response  = await client.get(only_url, params = params_value)

    data = response.json()
    

    return data
```

```python
result = await get_weather()
print(result)
```

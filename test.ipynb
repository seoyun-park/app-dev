```python
from pydantic import BaseModel

class PydanticUser(BaseModel):
    name:str

user2 = PydanticUser(name = 'baby')
user2.name, type(user2.name)
```

```python
class UserRegister(BaseModel):
    user_id : int
    username : str 
    email : str 
    is_admin : bool = False
```

```python
user = UserRegister(
    user_id=1,
    username="홍길동",
    email="hong@example.com"
)

user
```

```python
input_data = {
    "user_id": "1004",    # 문자열로 들어왔지만 int로 자동 변환(Parsing)됩니다.
    "username": "codemaster",
    "email": "code@example.com"
    # is_admin은 생략되었으므로 기본값 False가 적용됩니다.
}
```

```python
user1 = UserRegister(**input_data) #언팩킹
user1.user_id , type(user1.user_id)
```
